```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv(r"/Users/spoorthin/Downloads/archive/Iris.csv")
```

```python
# view first 5 rows
print("First 5 rows of the datasets:")
print(df.head())
```

```python
# view last 5 rows 
print("last 5 rows of the datasets:")
print(df.tail())
```

```python
#check shape
print(df.shape)
```

```python
#view column names
print(df.columns)
```

```python
#check info
print(df.info())
```

```python
#checking missing values
print(df.isnull().sum())
```

```python
#remove missing values and check the missing values again 
df=(df.dropna())
print(df.isnull().sum())
```

```python
#ckeck for duplicate rows
print(df.duplicated().sum())
```

```python
#if any duplicated rows.. remove it and check again
df=df.drop_duplicates()
print(df.duplicated().sum())
```

```python
#display statistical summary
print(df.describe())
```

```python
#set graph style
sns.set_style("whitegrid")
```

```python
#histogram plot
#create graph area
plt.figure(figsize=(8,5))
#create histogram
sns.histplot(data=df,x="SepalLengthCm",bins=10)
#add title
plt.title("Disturbution of sepal length")
#add x-axis
plt.xlabel("Sepal length (cm)")
#add y axis
plt.ylabel("Number of flowers")
#display
plt.show()
```

```python
#box plot
#create graph area
plt.figure(figsize=(8,5))
#create box plot
sns.boxplot(data=df,x="Species",y="SepalLengthCm")
#add title
plt.title("Sepal length by Species")
#add x-axis
plt.xlabel("Species")
#add y axis
plt.ylabel("Sepal length (cm)")
#display
plt.show()
```

```python
#scatter plot
#create graph area
plt.figure(figsize=(8,5))
#create box plot
sns.scatterplot(data=df,x="SepalLengthCm",y="PetalLengthCm",hue="Species")
#add title
plt.title("Sepal length vs Petal length")
#add x-axis
plt.xlabel("Sepal length (cm)")
#add y axis
plt.ylabel("Petal length (cm)")
#display
plt.show()
```

```python
#pair plot
#create pair plot
sns.pairplot(df,hue="Species")
#display graph
plt.show()
```

```python
#correlation
#select numerical column
numeric_columns=["SepalLengthCm","SepalWidthCm","PetalLengthCm","PetalWidthCm"]
#calculate correlation
correlation=df[numeric_columns].corr()
#display correlation matrix
print(correlation)
#heat map
#graph area
plt.figure(figsize=(10,5))
#create correlation heatmap
sns.heatmap(correlation,annot=True,cmap="coolwarm",fmt=".2f")
#title
plt.title("Correlation heatmap")
#display
plt.show()

```
